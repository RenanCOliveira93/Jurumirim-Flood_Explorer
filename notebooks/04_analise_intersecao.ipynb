```python
from jurumirim import data, gis
env = data.envelope(); mun = data.municipios(); urb = data.urbanizadas()
print(mun.crs.is_projected, 'envoltória km²:', round(env.area/1e6,2))
mo = gis.municipality_overlap(env, mun); mo.round(2)
```

```python
uo = gis.urban_overlap(env, urb, mun); uo.head(10).round(2)
```

```python
pir_m = mun[mun.codarea == gis.PIRAJU_CODE].geometry.iloc[0]
in_pir = env.intersection(pir_m)
print(f'Envoltória dentro de Piraju: {in_pir.area/1e6:.1f} km² ({100*in_pir.area/pir_m.area:.1f}% do município)')
us = data.urban_summary()
print(f"Maior polígono urbanizado de Piraju: {us['main_area_km2']:.2f} km²; distância à envoltória {us['main_distance_m']:.0f} m; intersecta: {us['main_intersects']}")
print(f"Polígonos urbanizados de Piraju que intersectam a envoltória: {us['n_intersecting']} (total {us['intersecting_area_km2']*1e6:.0f} m²)")
pd.DataFrame(us['intersecting_polygons'])
```

```python
rc = gis.road_crossings(env, data.roads()); rc
```

## Leitura
- Na digitalização, o **maior polígono urbanizado IBGE 2019 de Piraju não intersecta a envoltória** (~2 km de distância); **2 polígonos pequenos e pouco densos intersectam marginalmente** (centenas a poucos milhares de m², ≤1,1% de sua área) — ordem de grandeza da incerteza de digitalização. O PAE rotula ST-014/ST-015 como 'Área urbana Pirajú ME'.
- A parte da envoltória dentro de Piraju é rural/ribeirinha (represa da UHE Piraju e entorno, propriedades — PAE p. 52).
- A rodovia **SP-261** cruza a envoltória próximo à barragem (OSM).
- Isso **não** é classificação de segurança de endereços.
